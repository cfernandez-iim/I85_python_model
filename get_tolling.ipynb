```python
import pandas as pd
import numpy as np

lights_w = 1

heavies_w = 3
heavies_w_toll = 5
heavies_w_vot = 5

medium_A_w = 3 #1.5 # TBD: Maybe try 2.5 or 2.75 for every pce value
medium_A_w_toll = 5
medium_A_w_vot = 5

medium_B_w = 3 #2.75
medium_B_w_toll = 5
medium_B_w_vot = 5

heavy_A_w = 3 # 2.75
heavy_A_w_toll = 5
heavy_A_w_vot = 5

heavy_B_w = 3
heavy_B_w_toll = 5
heavy_B_w_vot = 5

def generate_revenue_stream(file_name, both_dirs):

    if both_dirs:
        directions = 2
    else:
        directions = 1

    output_df = pd.DataFrame(columns=['Year', 'Total Revenue', 'Total Transactions', 'AADT', 'Toll/AADT/mi']) # AADT refers to ML

    years = [2025, 2032, 2040, 2050]

    # years = [2032]

    Anualization_factor = 308 # TBD: Just changed from 290, check if it makes sense # Try 308
    traffic_anualization = 315 # TBD: Check if it fits
    year_days = 365
    length_correction_factor = 1.038

    # DataFrame with Tolling Segment as index and all other columns (except Length)
    first_model_df = pd.read_csv('test_get_toll/COLA_sup_new_v7/test_model_run.csv')

    tolling_segment = pd.read_excel('inputs/Segment to Tolling Segment Extensions.xlsx')

    df_length = tolling_segment[['Tolling Segment', 'Length']].set_index('Tolling Segment')

    # DataFrame with Tolling Segment as index and all other columns (except Length)
    df_percentages = tolling_segment.drop('Length', axis=1).set_index('Tolling Segment')

    phase_1_total_length = df_length[df_length.index.str.startswith('TS1')]['Length'].sum()
    phase_2_total_length = df_length[df_length.index.str.startswith('TS2')]['Length'].sum()
    total_length = df_length['Length'].sum()

    periods = ['Night', 'AM-Early', 'AM-Peak', 'AM-Shoulder', 'MD', 'PM-Shoulder', 'PM-Peak', 'PM-Late']

    for year in years:
        rows = []
        for index in df_percentages.index:
            for period in periods:

                ml_lights = 0
                ml_mediumA = 0
                ml_mediumB = 0
                revenue_mile = 0
                traffic_mile = 0

                count = 0

                toll = 0

                for column in df_percentages.columns:

                    try:
                        filtered_row = first_model_df[
                            (first_model_df["Period"] == period) &
                            (first_model_df["Year"] == year) &
                            (first_model_df["SegDir"] == column)
                        ].iloc[0]
                        
                    except IndexError:
                        print(f'Period: {period}, Year: {year}, SegDir: {column}')

                    toll = filtered_row["TollLights"]

                    if df_percentages.loc[index, column] != 0:

                        count += filtered_row["Length"]

                        ml_lights_aux = df_percentages.loc[index, column] * filtered_row["MLVeh_Lights"]
                        ml_mediumA_aux = df_percentages.loc[index, column] * filtered_row["MLVeh_MediumA"]
                        ml_mediumB_aux = df_percentages.loc[index, column] * filtered_row["MLVeh_MediumB"]

                        ml_lights += ml_lights_aux
                        ml_mediumA += ml_mediumA_aux
                        ml_mediumB += ml_mediumB_aux

                        revenue_mile += filtered_row["Length"] * (ml_lights_aux * toll + ml_mediumA_aux * toll * medium_A_w_toll + ml_mediumB_aux * toll * medium_B_w_toll) * filtered_row["Hours/Day"]
                        traffic_mile += df_percentages.loc[index,column] * filtered_row["TransactionsDay"] * filtered_row["Length"]
                
                revenue_mile = revenue_mile / count

                traffic_mile = traffic_mile / count

                rows.append({
                    "SegDir": index,
                    "Year": year,
                    "Length": df_length.loc[index, 'Length'],
                    "Period": period,
                    "Hours/Day": filtered_row["Hours/Day"],
                    "MLVeh_Lights": ml_lights,
                    "MLVeh_MediumA": ml_mediumA,
                    "MLVeh_MediumB": ml_mediumB,
                    "Revenue Per Mile": revenue_mile,
                    "Traffic": traffic_mile,
                    "TrafficLength": traffic_mile * df_length.loc[index, 'Length']
                })

        tolling_segment_df = pd.DataFrame(rows).set_index('SegDir')

        tolling_segment_df["RevenuePerDay"] = tolling_segment_df.apply(
            lambda row: row["Length"] * row["Revenue Per Mile"],
            axis=1
        )

        phase1_df = tolling_segment_df[tolling_segment_df.index.str.startswith('TS1')]
        phase2_df = tolling_segment_df[tolling_segment_df.index.str.startswith('TS2')]

        phase_1_revenues = phase1_df.loc[:,"RevenuePerDay"].sum() * Anualization_factor * length_correction_factor
        phase_2_revenues = phase2_df.loc[:,"RevenuePerDay"].sum() * Anualization_factor * length_correction_factor

        phase_12_revenues = tolling_segment_df.loc[:,"RevenuePerDay"].sum() * Anualization_factor * length_correction_factor

        phase_1_aadt = phase1_df.loc[:,"TrafficLength"].sum() / phase_1_total_length
        phase_2_aadt = phase2_df.loc[:,"TrafficLength"].sum() / phase_2_total_length

        phase_12_aadt = tolling_segment_df.loc[:,"TrafficLength"].sum() / (total_length / 2)

        traffic_1  = phase1_df.loc[:,"Traffic"].sum() * traffic_anualization
        traffic_2  = phase2_df.loc[:,"Traffic"].sum() * traffic_anualization
        traffic_12 = tolling_segment_df.loc[:,"Traffic"].sum() * traffic_anualization
        
        toll_AADT_mi_1 = phase_1_revenues / (year_days * phase_1_aadt * phase_1_total_length)
        toll_AADT_mi_12 = phase_12_revenues / (year_days * phase_12_aadt * total_length)

        print(f'Year: {year}')
        print(f'Phase 1 = {phase_1_revenues}, Phase 2 = {phase_2_revenues}, Phase 1 + 2 = {phase_1_revenues + phase_2_revenues}')
        print(f'Total Transactions = {traffic_1}, AADT = {phase_1_aadt}')

        tolling_segment_df.to_csv(f'tolling_segment_translated_{year}.csv')
        
        output_df.loc[len(output_df)] = [year, phase_12_revenues, traffic_12, phase_12_aadt, toll_AADT_mi_12]

    # Define full range of years to interpolate over
    full_years = pd.DataFrame({"Year": np.arange(int(output_df["Year"].min()), int(output_df["Year"].max()) + 1)})

    # Merge with original to create missing years with NaNs
    merged = pd.merge(full_years, output_df, on="Year", how="left")

    # Interpolate all numeric columns except "Year"
    interpolated = np.log(merged).interpolate(method="linear")

    interpolated = np.exp(interpolated)

    # Copy original column names
    original_columns = list(interpolated.columns)

    # Track how many columns we've inserted to adjust the index
    insert_count = 0

    # Start from index 1 to skip the first column (index 0)
    for i in range(1, len(original_columns)):
        col = original_columns[i]
        new_col_name = f"Var {col} (%)"
        insert_position = i + insert_count + 1  # Adjust position with insert_count
        interpolated[col] = interpolated[col].round(6)
        new_col_val = interpolated[col].pct_change()
        interpolated.insert(insert_position, new_col_name, new_col_val)
        insert_count += 1

    interpolated.to_csv(file_name)

    return interpolated
```
